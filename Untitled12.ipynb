```python
weeks = {
    "1" : "Monday",
    "2" : "Tuesday",
    "3" : "Wednesday",
    "4" : "Thursday",
    "5" : "Friday",
    "6" : "Saturday",
    "7" : "Sunday",
}
print(weeks)
print(weeks['1'])
print(weeks['3'])
print(weeks.get("9","Гараг нь зөвхөн 7 өдрөөс бүрдэнэ!"))
weeks['1'] = "Mon"
print(weeks)
for x in weeks:
  print(x)
for x in weeks.keys():
  print(x)
for x in weeks.values():
  print(x)
for x, y in weeks.items():
  print(x, y)
for d in weeks.items():
  print(d)


```
