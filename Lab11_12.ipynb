```python
# Алхам 1: input.txt файлыг үүсгэх (Жишээ оролт 1-ийг жишээ болгон оруулав. Та хүссэн оролт 1 эсвэл оролт 2-оо энд хуулж тавьж болно)
input_data = """Уртнасан Хүрэлбаатар Эмэгтэй БС АХС 96864108
Эрхбаяр Тогтох Эмэгтэй МБУС Т2 90889308
Бадамсүрэн Ренчин Эмэгтэй Лицей ГаТ 86223186
Гантогтох Бор Эрэгтэй АХБС Т2 80404926
Уламбаяр Ганбаяр Эмэгтэй Лицей Т3 86484938
Гармаа Сэсээр Эрэгтэй БС ОХТ 91539913
Баттогтох Сосор Эрэгтэй БТС Т3 99470642
Гоомарал Сүрэн Эрэгтэй БоСС Т4 95832190
Золзаяа Сэрээнэн Эмэгтэй ДУТС БиоТ 80634254
Эрхбаяр Батбаатар Эмэгтэй НХУС Т6 95674121"""

with open('input.txt', 'w', encoding='utf-8') as f:
    f.write(input_data)

# Алхам 2: Бодлогын үндсэн код
ifile = open('input.txt', 'r', encoding='utf-8')
mylines = ifile.readlines()
ifile.close()

teachers = []
salbar = {}
keys = ['ner', 'ovog', 'huis', 'salbar', 'tenhim', 'utas']

for line in mylines:
    values = line.split()
    if not values:
        continue
    rd = dict(zip(keys, values))

    # Салбар тус бүрээр тоолох
    if rd['salbar'] in salbar:
        salbar[rd['salbar']] += 1
    else:
        salbar[rd['salbar']] = 1

    teachers.append(rd)

# Гаралтын файлыг үүсгэх
ofile = open('output.txt', 'w', encoding='utf-8')
for row in sorted(salbar):
    ofile.write(row + ' ' + str(salbar[row]) + '\n')
ofile.close()

# Алхам 3: Үр дүнг дэлгэцэнд хэвлэж шалгах
print("--- output.txt файлын агуулга ---")
with open('output.txt', 'r', encoding='utf-8') as f:
    print(f.read())
```
